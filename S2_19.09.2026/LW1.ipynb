## 1. Списки (list)

**Список** — упорядоченная изменяемая коллекция. Самая универсальная структура для хранения последовательности значений: результатов измерений, строк таблицы, списка товаров.

Основные свойства:
- элементы нумеруются **с нуля**;
- поддерживаются **отрицательные индексы** (отсчёт с конца);
- **срезы** `a[start:stop:step]` возвращают новый список;
- список можно изменять: добавлять, удалять, заменять элементы.

````python
# Создание списков
sales = [120, 340, 560, 210, 480]        # выручка по дням
products = ['хлеб', 'молоко', 'сыр']     # список строк
mixed = [1, 'два', 3.0, True]            # можно смешивать типы
empty = []                               # пустой список

# Индексация и срезы
print(sales[0])       # первый элемент (индексация с нуля!)
print(sales[-1])      # последний элемент
print(sales[1:4])     # срез: индексы 1, 2, 3 (правая граница НЕ включается)
print(sales[::-1])    # список в обратном порядке
print(len(sales), sum(sales), max(sales))   # длина и агрегаты
````

````python
numbers = [5, 2, 8, 1, 9]

numbers.append(7)         # добавить элемент в конец
numbers.insert(0, 100)    # вставить 100 на позицию 0
numbers.remove(8)         # удалить первое вхождение значения 8
last = numbers.pop()      # извлечь и удалить последний элемент
numbers.sort()            # сортировка на месте (изменяет сам список!)
numbers.reverse()         # разворот на месте

print(numbers, '| извлечён элемент:', last)
print(100 in numbers, 999 not in numbers)    # проверка вхождения
print(sorted(numbers, reverse=True))         # sorted() возвращает НОВЫЙ список
````

## 2. Кортежи (tuple)

**Кортеж** — упорядоченная **неизменяемая** коллекция. После создания изменить его нельзя.

Когда использовать:
- данные, которые не должны меняться: координаты, дата, запись о человеке;
- возврат из функции нескольких значений;
- защита данных от случайного изменения.

````python
point = (55.75, 37.62)     # координаты Москвы: (широта, долгота)
print(point[0], point[1])

# Распаковка — очень частый приём в аналитике
lat, lon = point
print(f'Широта: {lat}, долгота: {lon}')

# Функции возвращают несколько значений — это тоже кортеж
q, r = divmod(17, 5)
print(f'17 // 5 = {q}, 17 % 5 = {r}')

# Кортеж из одного элемента требует запятой
print(type((42)), type((42,)))    # int и tuple — почувствуйте разницу

# Кортеж неизменяем — попытка изменения вызовет ошибку (раскомментируйте):
# point[0] = 0.0    # TypeError: 'tuple' object does not support item assignment
````

## 3. Множества (set)

**Множество** — неупорядоченная коллекция **уникальных** элементов. Дубликаты отбрасываются автоматически.

Множества идеальны, когда нужно:
- убрать дубликаты из данных;
- быстро проверять вхождение (`in` работает почти мгновенно даже на больших данных);
- находить пересечения, объединения и разности.

Основные операции: объединение `a | b`, пересечение `a & b`, разность `a - b`, симметрическая разность `a ^ b`.

````python
# Множество из списка — дубликаты исчезают
raw = ['spam', 'eggs', 'spam', 'ham', 'eggs', 'spam']
unique = set(raw)
print(unique, '| было', len(raw), 'стало', len(unique))

monday = {'ivan', 'maria', 'petr', 'anna'}
tuesday = {'maria', 'sergey', 'anna', 'olga'}

print('Приходили хотя бы раз:', monday | tuesday)   # объединение
print('Приходили оба дня:    ', monday & tuesday)   # пересечение
print('Новые во вторник:     ', tuesday - monday)   # разность
print('Только в один день:   ', monday ^ tuesday)   # симметрическая разность
# Обратите внимание: порядок вывода элементов множества не гарантирован.

print('anna' in monday, 'olga' in monday)   # быстрая проверка вхождения
````

## 4. Словари (dict)

**Словарь** — коллекция пар «ключ → значение». Главный инструмент структурирования данных в аналитике: записи о клиентах, параметры модели, частотные словари слов.

Важные свойства:
- ключи уникальны и неизменяемы (строки, числа, кортежи);
- доступ по ключу мгновенный;
- с Python 3.7 словарь помнит порядок добавления элементов.

````python
product = {
    'name': 'Ноутбук',
    'price': 54990,
    'in_stock': True,
    'tags': ['электроника', 'компьютеры'],
}

print(product['name'])              # доступ по ключу
print(product.get('rating'))        # ключа нет -> None (без ошибки!)
print(product.get('rating', 0))     # можно задать значение по умолчанию

product['price'] = 49990            # изменить значение по ключу
product['rating'] = 4.7             # добавить новую пару ключ-значение

for key, value in product.items():  # перебор пар
    print(f'{key}: {value}')

print(list(product.keys()))         # все ключи
print(list(product.values()))       # все значения
````

````python
# Классика аналитики: частотный словарь слов
text = 'яблоко банан яблоко груша банан яблоко'

counts = {}
for word in text.split():
    counts[word] = counts.get(word, 0) + 1   # get вернёт 0, если слова ещё нет

print(counts)
print('Самое частое слово:', max(counts, key=counts.get))
````

## 5. Функции: параметры и возвращаемые значения

**Функция** — именованный блок кода, который можно вызывать многократно. Это основа повторного использования кода.

```python
def имя_функции(параметры):
    """Документация: что делает функция."""
    тело
    return результат
```

Виды параметров:
- **позиционные** — передаются по порядку;
- **именованные (ключевые)** — передаются как `имя=значение`, порядок не важен;
- **со значением по умолчанию** — можно не указывать при вызове;
- `*args` — произвольное число позиционных аргументов;
- `**kwargs` — произвольное число именованных аргументов.

````python
def mean(values):
    """Среднее арифметическое последовательности чисел."""
    return sum(values) / len(values)


temperatures = [20, 22, 19, 25, 24]
print(f'Средняя температура: {mean(temperatures):.1f}')


def describe(values, name='ряд данных', digits=2):
    """Печатает сводку: имя, число элементов, сумма, среднее."""
    print(f'{name}: n={len(values)}, sum={sum(values)}, mean={mean(values):.{digits}f}')


describe(temperatures)                                  # значения по умолчанию
describe(temperatures, name='температуры', digits=1)    # именованные аргументы
````

````python
# *args — произвольное число позиционных аргументов
def min_max(*args):
    """Возвращает кортеж (минимум, максимум)."""
    return min(args), max(args)


lo, hi = min_max(4, 8, 15, 16, 23, 42)   # распаковка возвращённого кортежа
print(f'min = {lo}, max = {hi}')


# **kwargs — произвольное число именованных аргументов
def report(title, **kwargs):
    """Печатает заголовок и пары ключ-значение."""
    print(f'== {title} ==')
    for key, value in kwargs.items():
        print(f'  {key}: {value}')


report('Продажи за квартал', январь=100, февраль=150, март=130)


# Функция без return возвращает None
def no_return():
    x = 1

print(no_return())
````

**Важные правила:**
- функция без `return` возвращает `None`;
- `return` немедленно завершает работу функции;
- вернуть можно несколько значений через кортеж: `return a, b`;
- локальные переменные функции «не видны» снаружи — данные передавайте через параметры, а не через глобальные переменные.

### Задание 5. Функции

Работаем со списком `[3, 17, 8, 25, 6, 12, 25, 9]`.

1. Напишите функцию `median(values)` — медиану списка. Отсортируйте **копию** списка (подсказка: `sorted`) и возьмите средний элемент; если длина чётная — среднее двух центральных.
2. Напишите функцию `normalize(values)` — она возвращает **новый** список, где каждое значение поделено на максимум. Исходный список меняться не должен.
3. Напишите функцию `filter_above(values, threshold=10)` — возвращает список значений, строго больших `threshold`. Значение по умолчанию обязательно.
4. Проверьте все три функции на данных выше.

````python
# Решение задания 5
numbers = [3, 17, 8, 25, 6, 12, 25, 9]
def median(values):
    valuess = values.copy()
    valuess.sort()
    if len(valuess)%2==0:
        return valuess[(len(valuess)//2)-1], valuess[len(valuess)//2]
    else:
        return valuess[len(valuess)//2]
def normalize(values):
    new = []
    mx = max(values)
    for i in values:
        new.append(i/mx)
    return new
def filter_above(values, threshold=10):
    zn = []
    for i in values:
        if i>threshold:
            zn.append(i)
    return zn
print(median(numbers))
print(normalize(numbers))
print(filter_above(numbers))
    
````

## 6. Генераторы списков и функциональные приёмы обработки данных

**Генератор списка (list comprehension)** — компактный способ построить коллекцию из другой:

```python
[выражение for элемент in последовательность if условие]
```

Это декларативный стиль: мы описываем, **что** получить, а не шаги построения. В аналитическом коде такие конструкции встречаются постоянно.

````python
numbers = list(range(1, 11))

# Одно и то же двумя способами
squares = []
for n in numbers:                    # цикл: «как построить»
    squares.append(n ** 2)
squares_lc = [n ** 2 for n in numbers]   # comprehension: «что построить»

print(squares == squares_lc, squares_lc)

# Фильтрация: только чётные, в квадрате
print([n ** 2 for n in numbers if n % 2 == 0])

# Условие в выражении: if-else стоит ПЕРЕД for
print(['чёт' if n % 2 == 0 else 'нечёт' for n in numbers[:4]])

# Генератор словаря и генератор множества
prices = {'хлеб': 45, 'молоко': 89, 'сыр': 620}
print({name: price for name, price in prices.items() if price > 50})
print({p // 10 for p in prices.values()})
````

````python
sales = [('январь', 100), ('февраль', 150), ('март', 130)]

# lambda — «функция на месте», без имени
double = lambda x: x * 2
print(double(21))

# map применяет функцию к каждому элементу
amounts = list(map(lambda item: item[1], sales))
print(amounts)

# filter оставляет элементы, для которых функция вернула True
print(list(filter(lambda item: item[1] >= 130, sales)))

# sorted с key — сортировка по своему правилу
by_amount = sorted(sales, key=lambda item: item[1], reverse=True)
print('Лучший месяц:', by_amount[0])
print('max с key:  ', max(sales, key=lambda item: item[1]))

# zip — параллельный обход нескольких коллекций
months = ['янв', 'фев', 'мар']
plan = [110, 140, 150]
for month, fact, pl in zip(months, amounts, plan):
    print(f'{month}: факт {fact}, план {pl}, выполнение {fact / pl:.0%}')
````

**Comprehensions или map/filter?**

Оба стиля в Python равноправны. Генераторы списков обычно читаются легче, а `map`/`filter` удобны, когда функция уже написана. Правило хорошего тона: если логика сложнее одного выражения — выносите её в именованную функцию.

## 7. Декомпозиция решения на повторно используемые функции и модули

**Декомпозиция** — разбиение сложной задачи на простые подзадачи. Хорошая аналитическая программа — это набор небольших функций, каждая из которых решает одну задачу и имеет понятное имя.

Признаки хорошей функции:
- делает **одно** дело;
- имя описывает результат (`mean`, `filter_above`, `load_sales`), а не реализацию;
- получает данные через параметры и отдаёт результат через `return`;
- не печатает и не меняет внешние данные без необходимости — тогда её можно переиспользовать.

````python
def load_grades():
    """Загружает данные (в реальном проекте — чтение из файла или БД)."""
    return {
        'Аня': [5, 4, 5, 5],
        'Борис': [3, 4, 3],
        'Вера': [5, 5, 4, 5, 5],
        'Глеб': [4, 3, 4],
    }


def mean(values):
    """Среднее арифметическое."""
    return sum(values) / len(values)


def grade_averages(grades):
    """Средний балл каждого студента. Возвращает dict."""
    return {student: mean(grades[student]) for student in grades}


def top_students(averages, n=1):
    """Топ-n студентов по среднему баллу. Возвращает list[tuple]."""
    return sorted(averages.items(), key=lambda pair: pair[1], reverse=True)[:n]


# Конвейер: данные -> обработка -> результат
averages = grade_averages(load_grades())
print('Средние баллы:', averages)
print('Лучший студент:', top_students(averages))
````

### Модули: import

**Модуль** — файл с Python-кодом (`module.py`), который подключается командой `import`. Модули позволяют собирать функции в библиотеки и использовать их в разных проектах и ноутбуках.

```python
import math                    # весь модуль: math.sqrt(16)
import pandas as pd            # модуль под псевдонимом
from math import sqrt          # только одна функция
from statistics import mean    # можно вызывать без префикса
```

Важный приём — строка `if __name__ == '__main__':`. Код внутри неё выполняется **только** при запуске файла напрямую (`python module.py`) и не выполняется при импорте. Так один файл может быть и библиотекой функций, и исполняемым скриптом.

````python
%%writefile my_stats.py
"""my_stats — учебная библиотека описательных статистик."""


def mean(values):
    """Среднее арифметическое."""
    return sum(values) / len(values)


def variance(values):
    """Дисперсия."""
    m = mean(values)
    return sum((x - m) ** 2 for x in values) / len(values)


def describe(values, name='данные'):
    """Печатает краткую сводку по данным."""
    print(f'{name}: n={len(values)}, mean={mean(values):.2f}, var={variance(values):.2f}')


if __name__ == '__main__':
    # Выполнится только при прямом запуске файла: python my_stats.py
    print('Самопроверка:', mean([1, 2, 3]))
````

## 8. Итоговая мини-задача

Магазин ведёт учёт продаж. Каждый чек — кортеж `(номер чека, товар, количество, цена за штуку)`.

Реализуйте конвейер из функций и соберите отчёт:

1. `load_sales()` — возвращает данные (они заданы в ячейке ниже);
2. `revenue_per_item(sales)` — словарь «товар → суммарная выручка» (количество × цена по всем чекам);
3. `best_item(revenues)` — принимает результат п. 2 и возвращает пару «товар, выручка» с максимумом (используйте `max(..., key=...)`);
4. `items_report(sales)` — список пар «товар → выручка», отсортированный по убыванию выручки (`sorted` + `lambda`);
5. `total_revenue(sales)` — итоговая выручка магазина (одна строка: генератор списка + `sum`).

**Требования:** у каждой функции — докстринг; функции не печатают данные сами, печать результата — только в конце конвейера.

````python
# Данные для итоговой задачи
SALES = [
    (101, 'хлеб', 2, 45.0),
    (101, 'молоко', 1, 89.0),
    (102, 'молоко', 3, 89.0),
    (103, 'сыр', 1, 620.0),
    (104, 'хлеб', 1, 45.0),
    (105, 'сыр', 2, 620.0),
    (106, 'молоко', 2, 89.0),
    (107, 'йогурт', 4, 55.0),
]
def load_sales():
    """Загружает и возвращает исходный"""
    return [
        (101, 'хлеб', 2, 45.0),
        (101, 'молоко', 1, 89.0),
        (102, 'молоко', 3, 89.0),
        (103, 'сыр', 1, 620.0),
        (104, 'хлеб', 1, 45.0),
        (105, 'сыр', 2, 620.0),
        (106, 'молоко', 2, 89.0),
        (107, 'йогурт', 4, 55.0),
    ]

def revenue_per_item(sales):
    """Делает словать с общей выручкой по каждому товару"""
    slovar = {}
    for nm, name, count, price in sales:
       slovar[name] = slovar.get(name, count*price)+ count*price
    return slovar

def best_item(revenues):
    """Возвращает товар с максимальной выручкой и саму выручку"""
    a = max(revenues, key=revenues.get)
    return a, revenues[a]

def items_report(sales):
    """Выводит список пар(товар, выручка), отсортированный по убыванию выручки"""
    return sorted(sales.items() ,key=lambda x: x[1], reverse=True)

def total_revenue(sales):
    """Выводит итоговую выручку магазина"""
    return sum(list(x[1] for x in sales))

if __name__ == "__main__":
    data = load_sales()
    rev_per_item = revenue_per_item(data)
    top_item = best_item(rev_per_item)
    rep = items_report(rev_per_item)
    total = total_revenue(rep)
    print(f"Самый продаваемый товар: {top_item[0]} (Его выручка: {top_item[1]})")
    print(f"Рейтинг товаров по убыванию выручки: {rep}")
    print(f"Итоговая выручка магазина: {total}")
````

## ОТВЕТЫ НА ВОПРОСЫ
1. Список можно изменять, а кортеж после создания изменить нельзя. Кортеж нужен, когда данные должны быть неизменяемыми, например координаты точки.
2. Потому что set использует специальную структуру. Благодаря этому он быстро определяет, есть элемент или нет. А в списке элементы приходится проверять один за другим.
3. Она вернёт значение None. То есть фактически функция не возвращает никакого полезного значения.
4. append добавляет элемент в список, но сам ничего не возвращает. Поэтому a.append(5) добавит 5 в список, а при a = a.append(5) переменная a после этого станет равна None.
5. Если использовать квадратные скобки, и такого ключа нет, возникнет ошибка KeyError. Метод .get() ошибки не вызывает, а возвращает None или указанное значение по умолчанию.
6. Оба варианта применяют операцию к каждому элементу списка. Первый вариант обычно проще и понятнее читается. map удобнее, когда уже есть отдельная функция, которую нужно применить ко всем элементам.
7. Она нужна, чтобы определённая часть программы выполнялась только тогда, когда файл запускают непосредственно. Если этот файл импортировать в другую программу, этот код выполняться не будет.
8. Можно преобразовать список в set, а потом обратно в список. Но при таком способе может потеряться исходный порядок элементов.
